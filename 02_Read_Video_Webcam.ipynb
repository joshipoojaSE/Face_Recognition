# OpenCV: Read Video from a Webcam

In the previous notebook, you read a saved image using `cv2.imread()`. Now you'll capture **live images from a webcam** and display them continuously as a video.

**Key idea:** a video is a sequence of images (**frames**) displayed one after another.

**Workflow:** Connect to webcam (`cv2.VideoCapture()`) → Read a frame (`camera.read()`) → Display the frame (`cv2.imshow()`) → Repeat in a loop → Release the camera

## 1. Import OpenCV

We use `cv2` to access OpenCV's functions.

```python
import cv2
```

## 2. Connect to the webcam

- `cv2.VideoCapture()` creates an object to capture video.
- `0` usually represents the **default webcam**.
- `1` may represent another connected camera.

The camera ID depends on your system.

```python
camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("Could not open webcam")
```

## 3. Capture an image (frame)

`camera.read()` returns two values:

| Variable | Meaning |
|---|---|
| `success` | `True` if a frame was read successfully; otherwise `False` |
| `image` | The captured frame as an image array |

If the camera fails to provide a frame, we handle the error instead of trying to display an invalid image.

```python
success, image = camera.read()

if not success:
    print("Reading from camera failed")
else:
    print("Frame shape:", image.shape)
```

## 4. Display the frame

`cv2.imshow()` opens a window and displays the captured image.

To keep the window visible for a specified time, use `cv2.waitKey()`:
- `cv2.waitKey(4000)` waits **4,000 milliseconds** (4 seconds).
- `cv2.waitKey(0)` waits **indefinitely** for a key press.

```python
if success:
    cv2.imshow("Image Window", image)
    cv2.waitKey(4000)
    cv2.destroyAllWindows()

camera.release()
```

## 5. Why do we need a loop?

Capturing an image only once gives you **just one frame**, not a live video.

To display a continuous stream, repeatedly capture and display frames:

1. **Open webcam** — `cv2.VideoCapture(0)`
2. **Read a frame** — `camera.read()`
3. **Display the frame** — `cv2.imshow()`
4. **Repeat** continuously until the program is stopped.

## 6. Complete working example

The lecture's approach, with a keyboard shortcut to exit the loop and proper cleanup added.

**Why `waitKey(1)`?**
- `1` means a wait of approximately 1 millisecond.
- It also lets OpenCV process window events and keyboard input.
- The loop keeps capturing frames until you press **`q`**.

> The lecture uses **Ctrl+C** to stop the script from the terminal. This example instead lets you quit by pressing **`q`** in the OpenCV window.

```python
import cv2

camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("Could not open webcam")
else:
    while True:
        success, image = camera.read()

        if not success:
            print("Failed to read frame")
            break

        cv2.imshow("Webcam Video", image)

        # Exit when the user presses q
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    camera.release()
    cv2.destroyAllWindows()
```

## 7. Important functions to remember

| Function | Purpose |
|---|---|
| `cv2.VideoCapture(0)` | Connect to the webcam |
| `camera.read()` | Capture the next frame |
| `cv2.imshow()` | Display the frame |
| `cv2.waitKey(1)` | Process window events and check for a key |
| `camera.release()` | Release the webcam |
| `cv2.destroyAllWindows()` | Close OpenCV windows |
