# OpenCV Basics: Reading and Displaying Images

**Workflow:** Install OpenCV → Read image (`cv2.imread()`) → Convert BGR to RGB (`cv2.cvtColor()`) → Display image (`plt.imshow()`)

## 1. Install and import OpenCV

Install the library (run once). Although the library is called **OpenCV**, we import it as `cv2` in Python.

```python
pip install opencv-python matplotlib
```

```python
import cv2
import matplotlib.pyplot as plt
```

## 2. Read an image

Place an image named `dog.jpg` in the same folder as this notebook.

- `cv2.imread()` reads the image from the specified path.
- `image` stores the image as a **NumPy array** of pixel values.
- If the path is incorrect or the image cannot be read, `imread()` returns `None`.

```python
image = cv2.imread("Dog.jpg")

if image is None:
    raise FileNotFoundError("Could not read 'dog.jpg'. Check the file name and path.")
```

## 3. Understand the image shape

In the lecture, the shape is `(1197, 2000, 3)`:

| Value | Meaning |
|---|---|
| 1197 | Height: number of rows |
| 2000 | Width: number of columns |
| 3 | Three color channels |

The image has $1197 \times 2000 = 2{,}394{,}000$ pixels, and each pixel has three color values.

A color image can be represented as a 3D array:

$$\text{Image} = \text{Height} \times \text{Width} \times \text{Channels}$$

```python
print("Shape:", image.shape)
height, width, channels = image.shape
print(f"Height: {height}, Width: {width}, Channels: {channels}")
print("Total pixels:", height * width)
```

## 4. The three color channels

In a standard RGB image, each pixel has:
- **R** = Red
- **G** = Green
- **B** = Blue

For example, a pure red pixel is $[255, 0, 0]$: maximum red, zero green, zero blue (8-bit RGB).

> **Important:** OpenCV reads color images in **BGR** order, not RGB. So the first pixel returned by `cv2.imread()` contains blue, green, and red values in that order.

```python
# First pixel (top-left) in BGR order
b, g, r = image[0, 0]
print("image[0, 0] =", image[0, 0])
print(f"Blue: {b}, Green: {g}, Red: {r}")
```

## 5. Display the image (incorrect colors)

Matplotlib expects **RGB**, but OpenCV supplies **BGR**, so the colors will look wrong (e.g., blue and red are swapped).

```python
plt.imshow(image)
plt.title("BGR shown as RGB (wrong colors)")
plt.axis("off")
plt.show()
```

## 6. Convert BGR to RGB

`cv2.COLOR_BGR2RGB` tells OpenCV to rearrange the color channels from BGR order to RGB order.

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Converted to RGB (correct colors)")
plt.axis("off")
plt.show()
```

## 7. Complete workflow

1. **Install OpenCV** — `pip install opencv-python`
2. **Read image** — `cv2.imread()`
3. **Convert BGR to RGB** — `cv2.cvtColor()`
4. **Display image** — `plt.imshow()`

```python
# Full workflow in one cell
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("dog.jpg")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
